```python
# ADDING CLASSIFIERS TO EXISTING CODE 
import os
import torch
import torch.nn as nn
import torchvision.models as models
import torchvision.transforms as T
import torchvision.datasets as datasets
from torch.utils.data import DataLoader
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, confusion_matrix,
    precision_score, recall_score, roc_curve, auc,
    cohen_kappa_score, matthews_corrcoef, classification_report
)
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
import seaborn as sns

# -----------------------
# Device setup
# -----------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# -----------------------
# Build Backbone Function
# -----------------------
def build_backbone(name="resnet50"):
    if name == "resnet50":
        model = models.resnet50(weights="IMAGENET1K_V1")
        model.fc = nn.Identity()
    elif name == "mobilenetv2":
        model = models.mobilenet_v2(weights="IMAGENET1K_V1")
        model.classifier = nn.Identity()
    elif name == "vgg16":
        model = models.vgg16(weights="IMAGENET1K_V1")
        model.classifier = nn.Identity()
    else:
        raise ValueError("Unknown model")
    model.eval().to(device)

    transform = T.Compose([
        T.Resize((224,224)),
        T.ToTensor(),
        T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
    ])
    return model, transform

# -----------------------
# Extract Deep Features
# -----------------------
def deep_embeddings(loader, model):
    feats, labels = [], []
    with torch.no_grad():
        for imgs, lbls in tqdm(loader, desc="Extracting features", unit="batch"):
            imgs = imgs.to(device)
            emb = model(imgs).cpu().numpy()
            feats.append(emb)
            labels.append(lbls.numpy())
    return np.vstack(feats), np.hstack(labels)

# -----------------------
# Evaluation Function with 3 classifiers
# -----------------------
def evaluate_deep(X, y, name, use_pca=True, pca_dim=256, class_names=None, top_k=10):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )

    if use_pca:
        print(f"Applying PCA reduction to {pca_dim} dimensions...")
        pca = PCA(n_components=pca_dim, random_state=42)
        X_train = pca.fit_transform(X_train)
        X_test = pca.transform(X_test)

    classifiers = {
        "Logistic Regression": LogisticRegression(
            max_iter=2000, n_jobs=-1, solver="lbfgs", multi_class="multinomial"
        ),
        "Random Forest": RandomForestClassifier(
            n_estimators=300, max_depth=None, n_jobs=-1, random_state=42
        ),
        "SVM": SVC(kernel="linear", probability=True, random_state=42)
    }

    for clf_name, clf in classifiers.items():
        print(f"\nTraining {clf_name} for {name}...")
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        y_prob = clf.predict_proba(X_test)

        acc = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, average="macro")
        prec = precision_score(y_test, y_pred, average="macro", zero_division=0)
        rec = recall_score(y_test, y_pred, average="macro", zero_division=0)
        kappa = cohen_kappa_score(y_test, y_pred)
        mcc = matthews_corrcoef(y_test, y_pred)

        print(f"\n{name} + {clf_name} Results:")
        print(f"Accuracy     : {acc:.4f}")
        print(f"Precision    : {prec:.4f}")
        print(f"Recall       : {rec:.4f}")
        print(f"F1-macro     : {f1:.4f}")
        print(f"Cohen’s Kappa: {kappa:.4f}")
        print(f"MCC          : {mcc:.4f}")

        print("\nClassification Report:")
        print(classification_report(y_test, y_pred, target_names=class_names))

        # Confusion matrix
        cm = confusion_matrix(y_test, y_pred)
        counts = cm.sum(axis=1)
        top_idx = np.argsort(counts)[::-1][:top_k]
        cm_sub = cm[np.ix_(top_idx, top_idx)]
        labels_sub = [class_names[i] for i in top_idx]

        plt.figure(figsize=(12, 10))
        sns.heatmap(cm_sub, annot=True, fmt="d", cmap="Blues",
                    xticklabels=labels_sub, yticklabels=labels_sub)
        plt.xticks(rotation=90)
        plt.yticks(rotation=0)
        plt.title(f"{name} + {clf_name} Confusion Matrix (Top {top_k} classes)")
        plt.tight_layout()
        plt.show()

        # ROC curve
        classes = np.unique(y)
        y_test_bin = label_binarize(y_test, classes=classes)
        if y_test_bin.shape[1] > 1:
            fpr, tpr, roc_auc = {}, {}, {}
            for i, cls in enumerate(classes):
                fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
                roc_auc[i] = auc(fpr[i], tpr[i])

            fpr["micro"], tpr["micro"], _ = roc_curve(y_test_bin.ravel(), y_prob.ravel())
            roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

            plt.figure()
            plt.plot(fpr["micro"], tpr["micro"],
                     label=f"micro-average ROC (area = {roc_auc['micro']:.2f})",
                     color="navy", linestyle=":", linewidth=4)
            for i in range(min(5, len(classes))):
                plt.plot(fpr[i], tpr[i],
                         lw=2, label=f"Class {class_names[i]} (area = {roc_auc[i]:.2f})")
            plt.plot([0, 1], [0, 1], "k--", lw=2)
            plt.xlabel("False Positive Rate")
            plt.ylabel("True Positive Rate")
            plt.title(f"{name} + {clf_name} ROC Curve")
            plt.legend(loc="lower right")
            plt.show()

# -----------------------
# Visualize few features
# -----------------------
def visualize_features(dataset, model, n_samples=5):
    inv_normalize = T.Normalize(
        mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
        std=[1/0.229, 1/0.224, 1/0.225]
    )
    fig, axes = plt.subplots(2, n_samples, figsize=(15, 6))
    model.eval()
    for i in range(n_samples):
        img, lbl = dataset[i]
        with torch.no_grad():
            feat = model(img.unsqueeze(0).to(device)).cpu().numpy().flatten()
        axes[0, i].imshow(np.transpose(inv_normalize(img).numpy(), (1,2,0)))
        axes[0, i].set_title(f"Original: {dataset.classes[lbl]}")
        axes[0, i].axis("off")
        axes[1, i].plot(feat[:200])
        axes[1, i].set_title("Feat. vector slice")
    plt.tight_layout()
    plt.show()

# -----------------------
# Load Dataset & Run
# -----------------------
if __name__ == "__main__":
    DATASET_PATH = r"C:\Users\Adithya R\DL Assignment\256_ObjectCategories"

    _, base_transform = build_backbone("resnet50")
    dataset = datasets.ImageFolder(DATASET_PATH, transform=base_transform)
    loader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=0)

    print(f"Loaded {len(dataset)} images across {len(dataset.classes)} classes")
    class_names = dataset.classes

    # -----------------------
    # ResNet50
    # -----------------------
    resnet, _ = build_backbone("resnet50")
    res_feats, labels = deep_embeddings(loader, resnet)
    evaluate_deep(res_feats, labels, "ResNet50", class_names=class_names, top_k=10)
    visualize_features(dataset, resnet, n_samples=5)

    # -----------------------
    # MobileNetV2
    # -----------------------
    mobilenet, _ = build_backbone("mobilenetv2")
    mob_feats, labels = deep_embeddings(loader, mobilenet)
    evaluate_deep(mob_feats, labels, "MobileNetV2", class_names=class_names, top_k=10)
    visualize_features(dataset, mobilenet, n_samples=5)

    # -----------------------
    # VGG16
    # -----------------------
    vgg, _ = build_backbone("vgg16")
    vgg_feats, labels = deep_embeddings(loader, vgg)
    evaluate_deep(vgg_feats, labels, "VGG16", class_names=class_names, top_k=10)
    visualize_features(dataset, vgg, n_samples=5)

```

```python
import os
import torch
import torch.nn as nn
import torchvision.models as models
import torchvision.transforms as T
import torchvision.datasets as datasets
from torch.utils.data import DataLoader, Dataset
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    cohen_kappa_score, matthews_corrcoef
)
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
import numpy as np
from tqdm import tqdm

# -----------------------
# Device setup
# -----------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# -----------------------
# Build Backbone Function
# -----------------------
def build_backbone(name="resnet50"):
    if name == "resnet50":
        model = models.resnet50(weights="IMAGENET1K_V1")
        model.fc = nn.Identity()
    elif name == "mobilenetv2":
        model = models.mobilenet_v2(weights="IMAGENET1K_V1")
        model.classifier = nn.Identity()
    elif name == "vgg16":
        model = models.vgg16(weights="IMAGENET1K_V1")
        model.classifier = nn.Identity()
    else:
        raise ValueError("Unknown model")
    model.eval().to(device)

    transform = T.Compose([
        T.Resize((224,224)),
        T.ToTensor(),
        T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
    ])
    return model, transform

# -----------------------
# Extract Deep Features
# -----------------------
def deep_embeddings(loader, model):
    feats, labels = [], []
    with torch.no_grad():
        for imgs, lbls in tqdm(loader, desc="Extracting features", unit="batch"):
            imgs = imgs.to(device)
            emb = model(imgs).cpu().numpy()
            feats.append(emb)
            labels.append(lbls.numpy())
    return np.vstack(feats), np.hstack(labels)

# -----------------------
# Evaluation Function with 3 classifiers
# -----------------------
def evaluate_deep(X, y, name, use_pca=True, pca_dim=256):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )

    if use_pca:
        pca = PCA(n_components=pca_dim, random_state=42)
        X_train = pca.fit_transform(X_train)
        X_test = pca.transform(X_test)

    classifiers = {
        "LogReg": LogisticRegression(
            max_iter=2000, n_jobs=-1, solver="lbfgs", multi_class="multinomial"
        ),
        "RF": RandomForestClassifier(
            n_estimators=300, max_depth=None, n_jobs=-1, random_state=42
        ),
        "SVM": SVC(kernel="linear", probability=True, random_state=42)
    }

    results = {}
    for clf_name, clf in classifiers.items():
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)

        acc = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, average="macro")
        prec = precision_score(y_test, y_pred, average="macro", zero_division=0)
        rec = recall_score(y_test, y_pred, average="macro", zero_division=0)
        kappa = cohen_kappa_score(y_test, y_pred)
        mcc = matthews_corrcoef(y_test, y_pred)

        results[clf_name] = {
            "Accuracy": acc,
            "Precision": prec,
            "Recall": rec,
            "F1": f1,
            "Kappa": kappa,
            "MCC": mcc
        }
    return results

# -----------------------
# Gaussian Noise Dataset Wrapper
# -----------------------
class NoiseDataset(Dataset):
    def __init__(self, dataset, mean=0.0, std=0.1):
        self.dataset = dataset
        self.mean = mean
        self.std = std

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, idx):
        img, label = self.dataset[idx]
        noise = torch.randn_like(img) * self.std + self.mean
        noisy_img = torch.clamp(img + noise, 0., 1.)
        return noisy_img, label

# -----------------------
# Pretty print results
# -----------------------
def print_results(model_name, clean_results, noisy_results):
    print(f"\n===== {model_name} =====")
    print("{:<12} {:<10} {:<10} {:<10} {:<10} {:<10} {:<10}".format(
        "Classifier", "Dataset", "Acc", "Prec", "Rec", "F1", "Kappa", "MCC"
    ))
    for clf in clean_results.keys():
        cr = clean_results[clf]
        nr = noisy_results[clf]
        print("{:<12} {:<10} {:.4f} {:.4f} {:.4f} {:.4f} {:.4f} {:.4f}".format(
            clf, "Clean", cr["Accuracy"], cr["Precision"], cr["Recall"],
            cr["F1"], cr["Kappa"], cr["MCC"]
        ))
        print("{:<12} {:<10} {:.4f} {:.4f} {:.4f} {:.4f} {:.4f} {:.4f}".format(
            clf, "Noisy", nr["Accuracy"], nr["Precision"], nr["Recall"],
            nr["F1"], nr["Kappa"], nr["MCC"]
        ))

# -----------------------
# Load Dataset & Run
# -----------------------
if __name__ == "__main__":
    DATASET_PATH = r"C:\Users\Adithya R\DL Assignment\256_ObjectCategories"

    _, base_transform = build_backbone("resnet50")
    dataset = datasets.ImageFolder(DATASET_PATH, transform=base_transform)
    loader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=0)

    noisy_dataset = NoiseDataset(dataset, mean=0, std=0.1)
    noisy_loader = DataLoader(noisy_dataset, batch_size=64, shuffle=False, num_workers=0)

    class_names = dataset.classes

    # -----------------------
    # ResNet50
    # -----------------------
    resnet, _ = build_backbone("resnet50")
    res_feats, labels = deep_embeddings(loader, resnet)
    res_feats_noisy, labels_noisy = deep_embeddings(noisy_loader, resnet)
    res_clean_results = evaluate_deep(res_feats, labels, "ResNet50 Clean")
    res_noisy_results = evaluate_deep(res_feats_noisy, labels_noisy, "ResNet50 Noisy")
    print_results("ResNet50", res_clean_results, res_noisy_results)

    # -----------------------
    # MobileNetV2
    # -----------------------
    mobilenet, _ = build_backbone("mobilenetv2")
    mob_feats, labels = deep_embeddings(loader, mobilenet)
    mob_feats_noisy, labels_noisy = deep_embeddings(noisy_loader, mobilenet)
    mob_clean_results = evaluate_deep(mob_feats, labels, "MobileNetV2 Clean")
    mob_noisy_results = evaluate_deep(mob_feats_noisy, labels_noisy, "MobileNetV2 Noisy")
    print_results("MobileNetV2", mob_clean_results, mob_noisy_results)

    # -----------------------
    # VGG16
    # -----------------------
    vgg, _ = build_backbone("vgg16")
    vgg_feats, labels = deep_embeddings(loader, vgg)
    vgg_feats_noisy, labels_noisy = deep_embeddings(noisy_loader, vgg)
    vgg_clean_results = evaluate_deep(vgg_feats, labels, "VGG16 Clean")
    vgg_noisy_results = evaluate_deep(vgg_feats_noisy, labels_noisy, "VGG16 Noisy")
    print_results("VGG16", vgg_clean_results, vgg_noisy_results)
```

```python
# === Caltech-256 (trimmed to your 90 classes): classical features with full dataset ===
# Outputs go to: C:\Users\Adithya R\DL Assignment\main_output
# Installs (if needed): pip install numpy pandas scikit-image scikit-learn opencv-python-headless matplotlib tqdm

import os, glob, time
import numpy as np
import cv2
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, accuracy_score,
    precision_score, recall_score, f1_score,
    cohen_kappa_score, matthews_corrcoef
)

from skimage.feature import hog, local_binary_pattern
from skimage.color import rgb2gray
from skimage.filters import prewitt, sobel, scharr

# ---------------- CONFIG ----------------
DATA_DIR = r"C:\Users\Adithya R\DL Assignment\256_ObjectCategories"
OUT_DIR  = r"C:\Users\Adithya R\DL Assignment\90_cls_trad_main"
IMG_SIZE = (128, 128)     # modest size for classical features
TEST_SIZE = 0.2
RANDOM_SEED = 42
PCA_DIMS = 256
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(f"{OUT_DIR}/features", exist_ok=True)
os.makedirs(f"{OUT_DIR}/visuals/samples", exist_ok=True)
np.random.seed(RANDOM_SEED)

# ---------------- DATA LOADER ----------------
def load_dataset(root, img_size=(128,128)):
    cls_names = sorted([d for d in os.listdir(root) if os.path.isdir(os.path.join(root, d))])
    print(f"Found {len(cls_names)} classes in dataset folder.")

    imgs, labels, fnames = [], [], []
    label_map = {}
    for lbl, cls in enumerate(cls_names):
        label_map[lbl] = cls
        files = sorted(glob.glob(os.path.join(root, cls, "*.jpg")))
        for fp in files:
            im = cv2.imread(fp)
            if im is None: continue
            im = cv2.cvtColor(im, cv2.COLOR_BGR2RGB)
            im = cv2.resize(im, img_size, interpolation=cv2.INTER_AREA)
            imgs.append(im); labels.append(lbl); fnames.append(fp)

    pd.DataFrame.from_dict(label_map, orient="index", columns=["class_name"]).to_csv(f"{OUT_DIR}/label_map.csv")
    return np.asarray(imgs, np.uint8), np.asarray(labels, np.int32), np.asarray(fnames, object), label_map

print("Loading dataset...")
X, y, fnames, label_map = load_dataset(DATA_DIR, IMG_SIZE)
print(f"Loaded {len(X)} images across {len(np.unique(y))} classes.")

# ---------------- FEATURE EXTRACTORS ----------------
def extract_hog(batch):
    feats, vis_list = [], []
    for im in tqdm(batch, desc="HOG"):
        g = rgb2gray(im)
        v, hog_img = hog(
            g, orientations=9, pixels_per_cell=(8,8), cells_per_block=(2,2),
            block_norm="L2-Hys", transform_sqrt=True, visualize=True, feature_vector=True
        )
        feats.append(v); vis_list.append(hog_img)
    return np.asarray(feats), vis_list

def extract_lbp(batch, P=8, R=1):
    feats, maps = [], []
    for im in tqdm(batch, desc="LBP"):
        g = rgb2gray(im)
        g8 = (g * 255.0).astype("uint8")
        lbp = local_binary_pattern(g8, P=P, R=R, method="uniform")
        hist, _ = np.histogram(lbp.ravel(), bins=np.arange(0, P+3), range=(0, P+2))
        hist = hist.astype("float32"); hist /= (hist.sum() + 1e-8)
        feats.append(hist); maps.append(lbp)
    return np.asarray(feats), maps

def extract_edges(batch, method="canny"):
    feats, maps = [], []
    for im in tqdm(batch, desc=f"Edges-{method}"):
        g = cv2.cvtColor(im, cv2.COLOR_RGB2GRAY)
        if method == "canny":
            e = cv2.Canny(g, 100, 200)
        elif method == "prewitt":
            e = (prewitt(g) * 255).astype("uint8")
        elif method == "sobel":
            e = (sobel(g) * 255).astype("uint8")
        elif method == "scharr":
            e = (scharr(g) * 255).astype("uint8")
        elif method == "laplacian":
            e = cv2.Laplacian(g, cv2.CV_64F)
            e = np.uint8(np.absolute(e))
        else:
            raise ValueError("Unknown edge method")
        feats.append(e.flatten().astype(np.float32)/255.0); maps.append(e)
    return np.asarray(feats), maps

# ---------------- EXTRACT & SAVE ----------------
hog_feats, hog_vis   = extract_hog(X);               np.save(f"{OUT_DIR}/features/hog.npy", hog_feats)
lbp_feats, lbp_maps  = extract_lbp(X);               np.save(f"{OUT_DIR}/features/lbp.npy", lbp_feats)
canny_feats, ca_maps = extract_edges(X, "canny");    np.save(f"{OUT_DIR}/features/canny.npy", canny_feats)
prew_feats, pr_maps  = extract_edges(X, "prewitt");  np.save(f"{OUT_DIR}/features/prewitt.npy", prew_feats)
sobel_feats, so_maps = extract_edges(X, "sobel");    np.save(f"{OUT_DIR}/features/sobel.npy", sobel_feats)
scharr_feats, sc_maps= extract_edges(X, "scharr");   np.save(f"{OUT_DIR}/features/scharr.npy", scharr_feats)
lap_feats, la_maps   = extract_edges(X, "laplacian");np.save(f"{OUT_DIR}/features/laplacian.npy", lap_feats)
np.save(f"{OUT_DIR}/features/labels.npy", y)

# ---------------- SAMPLE VISUALS ----------------
picked, seen = [], set()
for i, lbl in enumerate(y):
    if lbl not in seen:
        picked.append(i); seen.add(lbl)
    if len(picked) == 5: break

for i in picked:
    cls = label_map[int(y[i])]
    fig = plt.figure(figsize=(12,8))
    plt.subplot(2,3,1); plt.imshow(X[i]);          plt.title(f"Original [{cls}]"); plt.axis("off")
    plt.subplot(2,3,2); plt.imshow(ca_maps[i], cmap="gray"); plt.title("Canny"); plt.axis("off")
    plt.subplot(2,3,3); plt.imshow(pr_maps[i], cmap="gray"); plt.title("Prewitt"); plt.axis("off")
    plt.subplot(2,3,4); plt.imshow(so_maps[i], cmap="gray"); plt.title("Sobel"); plt.axis("off")
    plt.subplot(2,3,5); plt.imshow(sc_maps[i], cmap="gray"); plt.title("Scharr"); plt.axis("off")
    plt.subplot(2,3,6); plt.imshow(la_maps[i], cmap="gray"); plt.title("Laplacian"); plt.axis("off")
    plt.tight_layout()
    plt.savefig(f"{OUT_DIR}/visuals/samples/{cls}_edges.png", dpi=150)
    plt.close(fig)

print(f"[saved] sample edge maps for {len(picked)} classes -> {OUT_DIR}\\visuals\\samples")

# ---------------- CLASSIFIER RUNNER ----------------
def run_classifiers(X, y, name, use_pca=False, pca_dims=256):
    Z = StandardScaler(with_mean=True, with_std=True).fit_transform(X)
    if use_pca and Z.shape[1] > pca_dims:
        Z = PCA(n_components=pca_dims, random_state=RANDOM_SEED, svd_solver="randomized").fit_transform(Z)

    X_tr, X_te, y_tr, y_te = train_test_split(
        Z, y, test_size=TEST_SIZE, stratify=y, random_state=RANDOM_SEED
    )

    models = {
        "LogReg": LogisticRegression(max_iter=4000, solver="saga", n_jobs=-1),
        "SVM(RBF)": SVC(kernel="rbf", C=10, gamma="scale"),
        "RandomForest": RandomForestClassifier(n_estimators=300, random_state=RANDOM_SEED, n_jobs=-1)
    }

    accs = {}
    for mname, clf in models.items():
        t0 = time.time()
        clf.fit(X_tr, y_tr)
        y_pr = clf.predict(X_te)

        acc   = accuracy_score(y_te, y_pr)
        prec  = precision_score(y_te, y_pr, average="macro", zero_division=0)
        rec   = recall_score(y_te, y_pr, average="macro", zero_division=0)
        f1    = f1_score(y_te, y_pr, average="macro", zero_division=0)
        kappa = cohen_kappa_score(y_te, y_pr)
        mcc   = matthews_corrcoef(y_te, y_pr)

        accs[mname] = acc
        rep_txt = classification_report(y_te, y_pr, zero_division=0)
        with open(f"{OUT_DIR}/{name}_{mname}_report.txt", "w", encoding="utf-8") as f:
            f.write(rep_txt)
            f.write("\n\n")
            f.write(f"Accuracy     : {acc:.4f}\n")
            f.write(f"Precision    : {prec:.4f}\n")
            f.write(f"Recall       : {rec:.4f}\n")
            f.write(f"F1-score     : {f1:.4f}\n")
            f.write(f"Cohen’s Kappa: {kappa:.4f}\n")
            f.write(f"MCC          : {mcc:.4f}\n")
            f.write(f"Time(s): {time.time()-t0:.1f}\n")

        print(f"{name} | {mname}: Acc={acc:.4f}, Prec={prec:.4f}, Rec={rec:.4f}, F1={f1:.4f}, Kappa={kappa:.4f}, MCC={mcc:.4f}, Time={time.time()-t0:.1f}s")

    return accs

print("\n=== Training 3 classifiers on each feature set ===")
acc_hog   = run_classifiers(hog_feats,  y, name="HOG",        use_pca=True,  pca_dims=PCA_DIMS)
acc_lbp   = run_classifiers(lbp_feats,  y, name="LBP",        use_pca=False)
acc_canny = run_classifiers(canny_feats,y, name="CannyEdges", use_pca=True,  pca_dims=PCA_DIMS)
acc_prew  = run_classifiers(prew_feats, y, name="Prewitt",    use_pca=True,  pca_dims=PCA_DIMS)
acc_sobel = run_classifiers(sobel_feats,y, name="Sobel",      use_pca=True,  pca_dims=PCA_DIMS)
acc_scharr= run_classifiers(scharr_feats,y,name="Scharr",     use_pca=True,  pca_dims=PCA_DIMS)
acc_lap   = run_classifiers(lap_feats, y,  name="Laplacian",  use_pca=True,  pca_dims=PCA_DIMS)

# ---------------- GROUPED BAR CHART ----------------
df = pd.DataFrame([
    {"feature":"HOG",       **acc_hog},
    {"feature":"LBP",       **acc_lbp},
    {"feature":"Canny",     **acc_canny},
    {"feature":"Prewitt",   **acc_prew},
    {"feature":"Sobel",     **acc_sobel},
    {"feature":"Scharr",    **acc_scharr},
    {"feature":"Laplacian", **acc_lap},
])
df = df.set_index("feature")

plt.figure(figsize=(10,6))
df.plot(kind="bar")
plt.ylabel("Accuracy")
plt.title("Classifier Accuracy by Feature Type")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/visuals/classifiers_by_feature.png", dpi=150)
plt.show()

print("\nAll outputs saved to:", OUT_DIR)
```
